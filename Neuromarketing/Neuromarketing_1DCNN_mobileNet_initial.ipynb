```python
def clean_time(val):
    if pd.isna(val):
        return np.nan

    # Excel serial date
    if val > 1000:
        return (val % 1) * 24

    # Excel fraction
    if 0 <= val < 1:
        return val * 24

    # HH.MM
    h = int(val)
    m = int(round((val - h) * 100))
    
    if m >= 60:
        return np.nan
    
    return h + m/60

dataset_df['TIME_clean'] = dataset_df['TIME'].apply(clean_time)
```

```python
import numpy as np

# Step 1: valid range
valid_mask = (
    (dataset_df['duration_sec'] >= 30) &
    (dataset_df['duration_sec'] <= 300)
)

# Step 2: global median
global_median = dataset_df.loc[valid_mask, 'duration_sec'].median()

# Step 3: define fix function
def fix_group(g):
    d = g.copy()

    # mark invalid as NaN
    d[(d < 30) | (d > 300)] = np.nan

    # fill within participant
    d = d.ffill().bfill()

    return d

# Step 4: IMPORTANT — use transform (not apply)
dataset_df['duration_sec_filled'] = dataset_df.groupby('SUBID')['duration_sec'].transform(fix_group)

# Step 5: fallback fill
dataset_df['duration_sec_filled'] = dataset_df['duration_sec_filled'].fillna(global_median)
```

```python
import numpy as np

sampling_rate = 32

X_signal = []
y = []

# --- STEP 1: compute scaling factor ---
total_required = (dataset_df['duration_sec_filled'] * sampling_rate).sum()
scale = len(signal_df) / total_required

print("Scaling factor:", scale)

# --- STEP 2: windowing ---
start = 0
cols = ['BVP','EDA','TEMP','X','Y','Z']

for _, row in dataset_df.iterrows():
    
    # scaled length
    length = max(50, int(row['duration_sec_filled'] * sampling_rate * scale))
    
    # avoid zero-length segments
    if length <= 0:
        continue
    
    end = start + length
    
    # last segment safety (fit exactly)
    if end > len(signal_df):
        end = len(signal_df)
    
    seg = signal_df[cols].iloc[start:end].values
    
    # skip empty segments
    if len(seg) == 0:
        continue
    
    # pad if too small
    if len(seg) < 500:
        pad = 500 - len(seg)
        seg = np.pad(seg, ((0, pad), (0, 0)), mode='edge')
    
    # resample to 500
    idx = np.linspace(0, len(seg) - 1, 500).astype(int)
    seg = seg[idx].astype(np.float32)
    
    X_signal.append(seg)
    y.append(row['label'])
    
    start = end

# --- STEP 3: convert ---
X_signal = np.array(X_signal, dtype=np.float32)
y = np.array(y, dtype=np.float32)

print("Shapes:", X_signal.shape, y.shape)
print("Used signal:", start, "/", len(signal_df))
```

```python
import os

base_path = '/kaggle/input/datasets/alicia2uu/neurobiosense-dataset/NeuroBioSense Dataset/NeuroBioSense/Advertisement Categories'

def find_video_path(subid, signal_id):
    
    subid = str(int(subid))
    ad = signal_id.split('_')[1][:3]   # A02
    
    for category in os.listdir(base_path):
        cat_path = os.path.join(base_path, category)
        
        if not os.path.isdir(cat_path):
            continue
        
        sub_path = os.path.join(cat_path, subid)
        if not os.path.isdir(sub_path):
            continue
        
        ad_path = os.path.join(sub_path, ad)
        if not os.path.isdir(ad_path):
            continue
        
        for root, _, files in os.walk(ad_path):
            for file in files:
                if file.endswith('.mp4'):
                    return os.path.join(root, file)
    
    return None
```

```python
import cv2
import random

def extract_frames(video_path, num_frames=10):
    
    cap = cv2.VideoCapture(video_path)
    frames = []
    
    total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    step = max(total // num_frames, 1)
    
    for i in range(0, total, step):
        cap.set(cv2.CAP_PROP_POS_FRAMES, i)
        ret, frame = cap.read()
        
        if not ret:
            break
        
        # resize first
        frame = cv2.resize(frame, (224, 224))
        
        #  AUGMENTATION (~40%)
        if random.random() < 0.4:
            frame = cv2.flip(frame, 1)  # horizontal flip
        
        if random.random() < 0.4:
            frame = cv2.convertScaleAbs(frame, alpha=1.1, beta=10)  # brightness
        
        # normalize LAST
        frame = frame / 255.0
        
        frames.append(frame)
        
        if len(frames) == num_frames:
            break
    
    cap.release()
    return frames
```

```python
# Biosignal input
signal_input = Input(shape=(500, 6))

x1 = Conv1D(32, 3, activation='relu')(signal_input)
x1 = MaxPooling1D(2)(x1)

x1 = Conv1D(64, 3, activation='relu')(x1)
x1 = MaxPooling1D(2)(x1)

x1 = Flatten()(x1)
x1 = Dense(128, activation='relu')(x1)
x1 = Dropout(0.5)(x1)

signal_features = Dense(64, activation='relu')(x1)

# Video input
video_input = Input(shape=(64,))

# Fusion
combined = Concatenate()([signal_features, video_input])

# Final layers
x = Dense(128, activation='relu')(combined)
x = Dropout(0.5)(x)

output = Dense(1, activation='sigmoid')(x)

# Model
model = Model(inputs=[signal_input, video_input], outputs=output)
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

checkpoint = ModelCheckpoint(
    '/kaggle/working/best_model.h5',
    monitor='val_loss',
    save_best_only=True,
    verbose=1
)

history = model.fit(
    [X_signal, X_video],
    y,
    epochs=30,
    batch_size=8,
    validation_split=0.2,
    callbacks=[early_stop, checkpoint]
)
```
